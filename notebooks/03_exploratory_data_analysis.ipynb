# 03. Exploratory Data Analysis

**Objective:** Explore distributions, central tendency, outliers and relationships within the developer survey data.

## 1. Load libraries and data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/LargeData/m2_survey_data.csv"
df = pd.read_csv(dataset_url)
df.head()
```

## 2. Inspect the dataset

```python
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]:,}")
```

```python
df.columns.tolist()
```

## 3. Age analysis

```python
median_age = df["Age"].median()
print(f"Median age: {median_age:.1f}")
```

```python
plt.figure(figsize=(10, 6))
df["Age"].value_counts().sort_index().plot(kind="bar")
plt.xlabel("Age")
plt.ylabel("Number of respondents")
plt.title("Age Distribution")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
df["Age"].plot(kind="box")
plt.ylabel("Age")
plt.title("Age Distribution: Box Plot")
plt.tight_layout()
plt.show()
```

## 4. Compensation analysis

```python
median_compensation = df["ConvertedComp"].median()
print(f"Median converted compensation: {median_compensation:,.2f}")
```

```python
q1 = df["ConvertedComp"].quantile(0.25)
q3 = df["ConvertedComp"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

df_no_outliers = df[df["ConvertedComp"].between(lower_bound, upper_bound, inclusive="both")].copy()
print(f"Rows before outlier filtering: {df["ConvertedComp"].notna().sum():,}")
print(f"Rows after IQR filtering: {len(df_no_outliers):,}")
```

```python
mean_compensation_without_outliers = df_no_outliers["ConvertedComp"].mean()
print(f"Mean compensation after IQR filtering: {mean_compensation_without_outliers:,.2f}")
```

## 5. Gender-specific compensation

```python
women = df[df["Gender"] == "Woman"]
median_women_compensation = women["ConvertedComp"].median()
print(f"Median converted compensation for women: {median_women_compensation:,.2f}")
```

## 6. Correlation checks

```python
correlations = df[["Respondent", "WorkWeekHrs", "CodeRevHrs", "CompTotal"]].corrwith(df["Age"])
correlations.sort_values(ascending=False)
```
