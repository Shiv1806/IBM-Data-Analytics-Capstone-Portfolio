# 05. Web Scraping

**Objective:** Extract programming-language salary information from a web page and save the resulting table as a CSV file for downstream analysis.

## 1. Import libraries

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## 2. Request the source page

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
response = requests.get(url, timeout=30)
response.raise_for_status()
html = response.text
```

## 3. Parse the HTML

```python
soup = BeautifulSoup(html, "html.parser")
table = soup.find("table")
if table is None:
    raise ValueError("Expected salary table was not found on the page.")
```

## 4. Inspect the scraped records

```python
rows = []
for row in table.find_all("tr"):
    cols = row.find_all("td")
    if len(cols) >= 4:
        language_name = cols[1].get_text(strip=True)
        average_annual_salary = cols[3].get_text(strip=True)
        rows.append({
            "Language": language_name,
            "Average Annual Salary": average_annual_salary
        })

scraped_preview = pd.DataFrame(rows)
scraped_preview.head(10)
```

## 5. Save the scraped dataset

```python
scraped_data = pd.read_html(str(table))[0]
scraped_data.to_csv("popular-languages.csv", index=False)
print(f"Saved {len(scraped_data):,} records to popular-languages.csv")
```

## 6. Validate the output

```python
output = pd.read_csv("popular-languages.csv")
print(output.shape)
output.head()
```

## Key takeaway

The workflow demonstrates a basic data acquisition pipeline: request a web page, parse its HTML structure, extract the required fields, and persist the result as a reusable CSV dataset.
