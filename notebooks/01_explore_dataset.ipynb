# 01. Data Exploration

**Objective:** Establish a clear understanding of the developer survey dataset before cleaning or analysis.

This notebook covers dataset loading, structure, dimensions, data types, average age and country coverage.

## 1. Import libraries

```python
import pandas as pd
```

## 2. Load the survey dataset

The dataset is hosted by IBM Skills Network. Loading it directly from the source keeps the notebook reproducible without storing a large raw dataset in the repository.

```python
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/LargeData/m1_survey_data.csv"
df = pd.read_csv(dataset_url)
```

## 3. Preview the data

```python
df.head(5)
```

## 4. Dataset dimensions

Check the number of observations and variables.

```python
rows, columns = df.shape
print(f"Rows: {rows:,}")
print(f"Columns: {columns:,}")
```

## 5. Data types

```python
df.dtypes
```

## 6. Basic descriptive checks

```python
mean_age = df["Age"].mean()
print(f"Mean age: {mean_age:.2f}")
```

```python
unique_countries = df["Country"].nunique()
print(f"Unique countries: {unique_countries:,}")
```

## Key takeaway

This initial exploration establishes the dataset structure and the main variables available for later cleaning and analysis. The next stage focuses on data quality and preparation.
